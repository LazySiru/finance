```python
def format_dashboard(df: pd.DataFrame) -> pd.DataFrame:
    """Format the screener DataFrame for display.

    Operates on the *display* column names from DISPLAY_COLUMN_NAME_MAPPING,
    so it's safe to call on any subset of columns.
    """
    out = df.copy()

    # --- helpers ---------------------------------------------------------
    def _abbrev(x, prefix=""):
        if pd.isna(x):
            return "N/A"
        x = float(x)
        sign = "-" if x < 0 else ""
        v = abs(x)
        for unit, scale in (("T", 1e12), ("B", 1e9), ("M", 1e6), ("K", 1e3)):
            if v >= scale:
                return f"{sign}{prefix}{v/scale:.2f}{unit}"
        return f"{sign}{prefix}{v:,.2f}"

    fmt_money_abbr = lambda x: _abbrev(x, prefix="$")          # $23.75B
    fmt_count_abbr = lambda x: _abbrev(x, prefix="")           # 29.31M
    fmt_currency   = lambda x: "N/A" if pd.isna(x) else f"${x:,.2f}"
    fmt_ratio      = lambda x: "N/A" if pd.isna(x) else f"{x:.2f}"
    fmt_pct        = lambda x: "N/A" if pd.isna(x) else f"{x*100:.2f}%"
    fmt_int        = lambda x: "N/A" if pd.isna(x) else f"{int(x):,}"
    fmt_str        = lambda x: "N/A" if pd.isna(x) else str(x).title()

    # --- column → formatter map -----------------------------------------
    formatters = {
        # money, abbreviated
        "Market Cap":     fmt_money_abbr,
        "Free Cashflow":  fmt_money_abbr,
        "Total Cash":     fmt_money_abbr,
        "Total Debt":     fmt_money_abbr,
        # count, abbreviated
        "Avg. Volume":    fmt_count_abbr,
        # prices
        "Price":          fmt_currency,
        "Forward EPS":    fmt_currency,
        "Target Price":   fmt_currency,
        # plain ratios (2 dp)
        "Forward PE":     fmt_ratio,
        "Trailing PE":    fmt_ratio,
        "PEG":            fmt_ratio,
        "EV/EBITDA":      fmt_ratio,
        "Price to Book":  fmt_ratio,
        "Debt to Equity": fmt_ratio,
        "Beta":           fmt_ratio,
        "Analyst Rating": fmt_ratio,
        # percentages (yfinance returns as decimal → ×100)
        "Revenue Growth":  fmt_pct,
        "Earnings Growth": fmt_pct,
        "Profit Margin":   fmt_pct,
        "OP Margin":       fmt_pct,
        "ROE":             fmt_pct,
        "Insr. Ownership": fmt_pct,
        "Short Float":     fmt_pct,
        "Dividend Yield":  fmt_pct,
        "Payout Ratio":    fmt_pct,
        # ints
        "Analyst Opinions": fmt_int,
        # strings
        "Name":           lambda x: "N/A" if pd.isna(x) else str(x),
        "Sector":         fmt_str,
        "Industry":       fmt_str,
        "Recommendation": fmt_str,
    }

    for col, fn in formatters.items():
        if col in out.columns:
            out[col] = out[col].apply(fn)

    return out
```

```python
# TODO : Optimize add_stocks to use tickers.
def add_stock(df: pd.DataFrame, ticker_symbol: str) -> pd.DataFrame:
    ticker = yf.Ticker(ticker_symbol)
    info = ticker.info

    row = {
        field: info.get(field, None)
        for field in IMPORTANT_FIELDS
    }

    new_row = pd.DataFrame([row])

    return pd.concat([df, new_row], ignore_index=True)

def add_stocks(df: pd.DataFrame, stock_tickers) -> pd.DataFrame:
    for ticker in stock_tickers:
        df = add_stock(df, ticker)
    return df


def abbreviateNumber(value: float):
    if pd.isna(value):
        return value

    abs_value = abs(value)

    if abs_value >= 1_000_000_000:
        return f"{value / 1_000_000_000:.2f}B"
    elif abs_value >= 1_000_000:
        return f"{value / 1_000_000:.2f}M"
    elif abs_value >= 1_000:
        return f"{value / 1_000:.2f}K"
    else:
        return f"{value:.2f}"

# Define column groups
financial_cols = ['marketCap', 'currentPrice', 'forwardEps', "freeCashflow", 'totalCash', "totalDebt", "freeCashflow"]

percentage_cols = [
    'revenueGrowth', 'earningsGrowth', 'profitMargins',
    'operatingMargins', 'returnOnEquity', 'heldPercentInstitutions', 'shortPercentOfFloat','trailingAnnualDividendYield',
]

def format_df(df: pd.DataFrame) -> pd.DataFrame:

    formatted_df = df.copy()

    # Rename columns
    formatted_df.rename(columns=DISPALY_COLUMN_NAME_MAPPING, inplace=True)

    formatted_df = format_dashboard(formatted_df)

    return formatted_df


    # Generate stocks dataframe
def generate_stocks_df(stock_tickers):

  # Create df w/ full IMPORTANT_FIELDS columns
  stocks = pd.DataFrame(columns=IMPORTANT_FIELDS)

  stocks = add_stocks(stocks, stock_tickers)

  # Reset index to Ticker
  stocks.set_index("symbol", inplace=True)

  # Select columns to be displayed
  stocks = stocks[SELECTED_FIELDS]

  # Format data - Display Format & Renaming Cols
  stocks = format_df(stocks)

  return stocks
```

### Ticker.info

#### 📊 Interpreting `yfinance.ticker.info` Fields for Stock Evaluation

This guide summarizes how professional investors interpret key `ticker.info` properties when evaluating a company for investment decisions.

---

### 🧠 1. Company Overview

| Field | Interpretation |
|---|---|
| `longName` / `shortName` | Company identity |
| `sector` | Macro classification (Technology, Healthcare, etc.) |
| `industry` | Direct competitors & valuation comparison group |
| `longBusinessSummary` | Business model understanding |
| `fullTimeEmployees` | Operational scale & maturity |

✅ Used for **business understanding + peer comparison**

---

### 💰 2. Valuation Metrics (Is the Stock Expensive?)

| Field | Interpretation |
|---|---|
| `marketCap` | Company size (risk profile & fund eligibility) |
| `enterpriseValue` | True company value including debt |
| `trailingPE` | Price relative to past earnings |
| `forwardPE` | Expected valuation based on forecasts |
| `priceToSalesTrailing12Months` | Useful when profits are volatile |
| `priceToBook` | Asset valuation efficiency |
| `pegRatio` | Valuation adjusted for growth |

✅ Goal: Determine **cheap vs overpriced**

**Typical Signals**
- Lower PE relative to growth → attractive
- PEG ≈ 1 → fairly valued
- Very high multiples → growth expectations priced in

---

### 📈 3. Growth Metrics (Future Potential)

| Field | Interpretation |
|---|---|
| `revenueGrowth` | Sales expansion rate |
| `earningsGrowth` | Profit growth sustainability |
| `earningsQuarterlyGrowth` | Near-term momentum |
| `forwardEps` | Expected profitability |
| `revenuePerShare` | Revenue efficiency |

✅ Growth drives long-term stock returns.

**Healthy Company**
- Revenue growth > 5–10%
- Earnings growth ≥ revenue growth

---

### 🏆 4. Profitability & Business Quality

| Field | Interpretation |
|---|---|
| `profitMargins` | Net profitability |
| `grossMargins` | Pricing power |
| `operatingMargins` | Operational efficiency |
| `returnOnAssets` | Asset utilization |
| `returnOnEquity` | Capital efficiency (elite metric) |
| `ebitdaMargins` | Core operating strength |

✅ High margins + high ROE = **quality company**

---

### 💵 5. Cash Flow & Financial Strength

| Field | Interpretation |
|---|---|
| `freeCashflow` | Real cash generated for investors |
| `operatingCashflow` | Cash from core operations |
| `totalCash` | Liquidity buffer |
| `totalDebt` | Financial obligations |
| `debtToEquity` | Leverage risk |
| `currentRatio` | Short-term solvency |
| `quickRatio` | Immediate liquidity health |

✅ Companies survive recessions through **cash flow strength**.

---

### ⚖️ 6. Risk & Volatility

| Field | Interpretation |
|---|---|
| `beta` | Market volatility vs S&P 500 |
| `overallRisk` | Governance risk assessment |
| `auditRisk` | Accounting reliability |
| `boardRisk` | Board independence quality |
| `compensationRisk` | Executive incentive alignment |

**Beta Guide**
- < 1 → defensive
- ≈ 1 → market-like
- > 1.5 → high volatility

---

### 🧲 7. Ownership & Market Sentiment

| Field | Interpretation |
|---|---|
| `heldPercentInstitutions` | Institutional confidence |
| `heldPercentInsiders` | Management alignment |
| `sharesShort` | Bearish positioning |
| `shortPercentOfFloat` | Short squeeze potential |
| `shortRatio` | Days needed to cover shorts |

✅ Institutions often signal **smart money interest**.

---

### 📊 8. Analyst Expectations

| Field | Interpretation |
|---|---|
| `targetMeanPrice` | Consensus valuation |
| `targetHighPrice` | Bull case |
| `targetLowPrice` | Bear case |
| `recommendationMean` | Analyst rating score |
| `numberOfAnalystOpinions` | Confidence level |

Typical Scale:
- 1 → Strong Buy
- 2 → Buy
- 3 → Hold
- 4 → Sell

---

### 📉 9. Trading & Liquidity Metrics

| Field | Interpretation |
|---|---|
| `averageVolume` | Liquidity for entry/exit |
| `volume` | Current trading interest |
| `fiftyDayAverage` | Medium trend |
| `twoHundredDayAverage` | Long-term trend |
| `52WeekHigh` / `52WeekLow` | Price positioning |

✅ Used for timing entries.

---

### 🏦 10. Balance Sheet Efficiency

| Field | Interpretation |

##### how many consecutive days of rise and the percentage in the SPY and QQQ from the historical data and create histogram of consecutives days of rise per bar

```python
import pandas as pd
import matplotlib.pyplot as plt

def get_consecutive_days(df):
    data = df.copy()

    # Daily return
    data['Return'] = data['Close'].pct_change()

    # Direction (+1 up, -1 down)
    data['Direction'] = data['Return'].apply(
        lambda x: 1 if x > 0 else (-1 if x < 0 else 0)
    )

    # Create streak groups
    data['Group'] = (
        data['Direction'] != data['Direction'].shift()
    ).cumsum()

    rows = []

    for _, group in data.groupby('Group'):

        direction = group['Direction'].iloc[0]

        # skip flat moves
        if direction == 0:
            continue

        start_date = group.index[0]
        end_date = group.index[-1]

        period = len(group)

        start_price = group['Close'].iloc[0]
        end_price = group['Close'].iloc[-1]

        movement = (end_price / start_price - 1) * 100

        rows.append({
            "Date Range": f"{start_date.date()} → {end_date.date()}",
            "Period": period,
            "Movement %": round(movement, 2)
        })

    result = pd.DataFrame(rows).set_index("Date Range")

    return result

```

```python
spy = yfinance.Ticker('SPY')
qqq = yfinance.Ticker('QQQ')

spy_df = spy.history(period='max')
qqq_df = qqq.history(period='max')

spy_streaks = get_consecutive_days(spy_df)
spy_streaks
```

```python
import numpy as np
import matplotlib.pyplot as plt

# compute adaptive thresholds
q1, q2, q3, q4 = spy_streaks["Movement %"].quantile(
    [0.2, 0.4, 0.6, 0.8]
)

def get_color(p_low, p_high):

    mask = (periods >= p_low) & (periods < p_high)
    subset = moves[mask]

    if len(subset) == 0:
        return "gray"

    avg_move = subset.mean()

    if avg_move < q1:
        return "darkred"
    elif avg_move < q2:
        return "red"
    elif avg_move < q3:
        return "gray"
    elif avg_move < q4:
        return "green"
    else:
        return "darkgreen"

periods = spy_streaks["Period"]

moves = spy_streaks["Movement %"]

counts, bins = np.histogram(periods, bins=20)

plt.figure()

for i in range(len(counts)):

    left = bins[i]

    width = bins[i+1] - bins[i]

    height = counts[i]

    color = get_color(bins[i], bins[i+1])

    plt.bar(left, height, width=width, align="edge", color=color)

plt.title("Distribution of Consecutive Days of Movement (SPY)")

plt.xlabel("Period")

plt.ylabel("Frequency")

plt.show()
```

```python
"""
Financial Dashboard — yfinance + Plotly
Run: pip install yfinance plotly pandas
     python financial_dashboard.py
"""

import yfinance as yf
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots


# ── Config ────────────────────────────────────────────────
TICKER  = "AAPL"
PERIOD  = "6mo"     # 1mo 3mo 6mo 1y 2y 5y
MA_DAYS = 20        # Moving average window
SHOW_MA  = True
SHOW_RSI = True
RSI_PERIOD = 14
# ─────────────────────────────────────────────────────────


def calc_ma(series: pd.Series, window: int) -> pd.Series:
    return series.rolling(window=window).mean()


def calc_rsi(series: pd.Series, period: int = 14) -> pd.Series:
    delta = series.diff()
    gain  = delta.clip(lower=0).rolling(period).mean()
    loss  = (-delta.clip(upper=0)).rolling(period).mean()
    rs    = gain / loss
    return 100 - (100 / (1 + rs))


def fmt_large(n):
    if n is None: return "N/A"
    if n >= 1e12: return f"${n/1e12:.2f}T"
    if n >= 1e9:  return f"${n/1e9:.2f}B"
    if n >= 1e6:  return f"${n/1e6:.2f}M"
    return f"${n:,.0f}"

def fmt(v, prefix="", suffix="", decimals=2):
    if v is None or (isinstance(v, float) and (v != v)):
        return "N/A"
    return f"{prefix}{v:.{decimals}f}{suffix}"


def fetch_data(ticker, period):
    print(f"\nFetching {ticker} ({period})...")
    t    = yf.Ticker(ticker)
    hist = t.history(period=period)
    info = t.info
    try:
        recs = t.recommendations
    except Exception:
        recs = None
    return hist, info, recs


def print_metrics(info):
    print("\n" + "="*52)
    print(f"  {info.get('shortName', TICKER)}  ({info.get('symbol', TICKER)})")
    print(f"  {info.get('sector','')} · {info.get('industry','')}")
    print("="*52)
    rows = [
        ("Current Price",    fmt(info.get("currentPrice"),   prefix="$")),
        ("Market Cap",       fmt_large(info.get("marketCap"))),
        ("P/E Ratio",        fmt(info.get("trailingPE"))),
        ("Forward P/E",      fmt(info.get("forwardPE"))),
        ("EPS (TTM)",        fmt(info.get("trailingEps"),    prefix="$")),
        ("Forward EPS",      fmt(info.get("forwardEps"),     prefix="$")),
        ("Price/Book",       fmt(info.get("priceToBook"))),
        ("Dividend Yield",   fmt(info.get("dividendYield"),  suffix="%", decimals=2) if info.get("dividendYield") else "N/A"),
        ("Beta",             fmt(info.get("beta"))),
        ("52w High",         fmt(info.get("fiftyTwoWeekHigh"), prefix="$")),
        ("52w Low",          fmt(info.get("fiftyTwoWeekLow"),  prefix="$")),
        ("Target Price",     fmt(info.get("targetMeanPrice"),  prefix="$")),
        ("Analyst Rating",   str(info.get("recommendationKey", "N/A")).upper()),
        ("# Analysts",       str(info.get("numberOfAnalystOpinions", "N/A"))),
        ("Return on Equity", fmt(info.get("returnOnEquity") and info.get("returnOnEquity")*100, suffix="%")),
        ("Profit Margin",    fmt(info.get("profitMargins")  and info.get("profitMargins")*100,  suffix="%")),
    ]
    for label, value in rows:
        print(f"  {label:<22} {value}")
    print("="*52)


def print_recommendations(recs):
    if recs is None or recs.empty:
        print("\nNo analyst recommendation data available.")
        return
    print("\nAnalyst Recommendations (recent):")
    print(f"  {'Period':<12} {'StrongBuy':>9} {'Buy':>6} {'Hold':>6} {'Sell':>6} {'StrongSell':>11}")
    print("  " + "-"*54)
    for _, row in recs.tail(6).iterrows():
        print(f"  {str(row.get('period','')):<12}"
              f"  {int(row.get('strongBuy',0)):>7}"
              f"  {int(row.get('buy',0)):>6}"
              f"  {int(row.get('hold',0)):>6}"
              f"  {int(row.get('sell',0)):>6}"
              f"  {int(row.get('strongSell',0)):>10}")


def build_plot(hist, ticker):
    rows = 2 if SHOW_RSI else 1
    row_heights = [0.7, 0.3] if SHOW_RSI else [1.0]

    fig = make_subplots(
        rows=rows, cols=1,
        shared_xaxes=True,
        row_heights=row_heights,
        vertical_spacing=0.04,
        subplot_titles=[f"{ticker} — Price", "RSI (14)"] if SHOW_RSI else [f"{ticker} — Price"],
    )

    # ── Close price ──
    fig.add_trace(go.Scatter(
        x=hist.index, y=hist["Close"].round(2),
        name="Close", line=dict(color="#3266ad", width=1.5),
        hovertemplate="%{x|%b %d, %Y}<br>Close: $%{y:.2f}<extra></extra>",
    ), row=1, col=1)

    # ── Moving average ──
    if SHOW_MA:
        ma = calc_ma(hist["Close"], MA_DAYS)
        fig.add_trace(go.Scatter(
            x=hist.index, y=ma.round(2),
            name=f"MA {MA_DAYS}d", line=dict(color="#e07d28", width=1.5, dash="dot"),
            hovertemplate=f"MA{MA_DAYS}: $%{{y:.2f}}<extra></extra>",
        ), row=1, col=1)

    # ── RSI ──
    if SHOW_RSI:
        rsi = calc_rsi(hist["Close"], RSI_PERIOD)
        fig.add_trace(go.Scatter(
            x=hist.index, y=rsi.round(2),
            name="RSI", line=dict(color="#a32d2d", width=1.5),
            hovertemplate="RSI: %{y:.1f}<extra></extra>",
        ), row=2, col=1)

        # Overbought / oversold bands
        for level, color, label in [(70, "rgba(163,45,45,0.08)", "Overbought"),
                                     (30, "rgba(59,109,17,0.08)", "Oversold")]:
            fig.add_hline(y=level, line=dict(color="gray", width=0.8, dash="dash"), row=2, col=1)
        fig.update_yaxes(range=[0, 100], row=2, col=1)

    fig.update_layout(
        title=f"{ticker} Financial Dashboard",
        height=620,
        hovermode="x unified",
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="left", x=0),
        xaxis_rangeslider_visible=False,
        plot_bgcolor="white",
        paper_bgcolor="white",
        font=dict(family="Arial", size=12),
    )
    fig.update_xaxes(showgrid=True, gridcolor="#e8e8e8")
    fig.update_yaxes(showgrid=True, gridcolor="#e8e8e8")

    return fig


def main():
    hist, info, recs = fetch_data(TICKER, PERIOD)

    if hist.empty:
        print(f"No price data returned for {TICKER}. Check the ticker symbol.")
        return

    print_metrics(info)
    print_recommendations(recs)

    fig = build_plot(hist, TICKER)
    fig.show()
    print("\nChart opened in browser.")


if __name__ == "__main__":
    main()
```

#### Scoring model - scratch 1

```python
FACTOR_FIELDS = {
    "profitability": [
        "returnOnEquity",
        "profitMargins",
        "operatingMargins"
    ],
    "growth": [
        "revenueGrowth",
        "earningsGrowth",
        "earningsQuarterlyGrowth"
    ],
    "valuation": [
        "forwardPE",
        "pegRatio",
        "enterpriseToEbitda"
    ],
    "financial_health": [
        "debtToEquity",
        "currentRatio",
        "freeCashflow"
    ],
    "sentiment": [
        "heldPercentInstitutions",
        "shortPercentOfFloat"
    ]
}
```

```python
def normalize(value, low, high, reverse=False):
    if value is None:
        return 0.5

    value = max(min(value, high), low)
    score = (value - low) / (high - low)

    return 1 - score if reverse else score
```

```python
def score_metric(key, value):

    rules = {
        # profitability (higher better)
        "returnOnEquity": (0, 0.40, False),
        "profitMargins": (0, 0.30, False),
        "operatingMargins": (0, 0.30, False),

        # growth
        "revenueGrowth": (-0.1, 0.40, False),
        "earningsGrowth": (-0.2, 0.50, False),
        "earningsQuarterlyGrowth": (-0.2, 0.60, False),

        # valuation (lower better)
        "forwardPE": (5, 60, True),
        "pegRatio": (0.5, 3, True),
        "enterpriseToEbitda": (5, 40, True),

        # financial health
        "debtToEquity": (0, 200, True),
        "currentRatio": (0.5, 3, False),
        "freeCashflow": (0, 20_000_000_000, False),

        # sentiment
        "heldPercentInstitutions": (0.2, 0.9, False),
        "shortPercentOfFloat": (0, 0.20, True),
    }

    if key not in rules:
        return 0.5

    low, high, reverse = rules[key]
    return normalize(value, low, high, reverse)
```

```python
def score_factor(info, fields):
    scores = []

    for f in fields:
        value = info.get(f)
        if value is not None:
            scores.append(score_metric(f, value))

    return sum(scores) / len(scores) if scores else 0.5
```

```python
FACTOR_WEIGHTS = {
    "profitability": 0.25,
    "growth": 0.25,
    "valuation": 0.20,
    "financial_health": 0.20,
    "sentiment": 0.10,
}


def quant_score(info):

    total = 0

    for factor, fields in FACTOR_FIELDS.items():
        factor_score = score_factor(info, fields)
        total += factor_score * FACTOR_WEIGHTS[factor]

    return round(total * 100, 2)
```

#### Scoring model - scratch 2

```python
def load_yf_data(symbol):

    t = yf.Ticker(symbol)

    try:
        return {
            "price": t.history(period="10y")["Close"],
            "income": t.quarterly_income_stmt,
            "balance": t.balance_sheet,
            "cashflow": t.cashflow,
            "info": t.info
        }
    except:
        return None
```

```python
def price_factors(price):

    returns = price.pct_change()

    momentum_12m = price.iloc[-1] / price.iloc[-252] - 1
    volatility = returns.std() * (252**0.5)

    return {
        "momentum_12m": momentum_12m,
        "volatility": volatility
    }

def income_factors(income):

    try:
        revenue = income.loc["Total Revenue"]
        net_income = income.loc["Net Income"]

        revenue_growth = revenue.pct_change().mean()
        earnings_growth = net_income.pct_change().mean()

        return {
            "rev_growth_q": revenue_growth,
            "earn_growth_q": earnings_growth
        }
    except:
        return {}


def balance_factors(balance):

    try:
        debt = balance.loc["Total Debt"].iloc[0]
        equity = balance.loc["Stockholders Equity"].iloc[0]
        cash = balance.loc["Cash And Cash Equivalents"].iloc[0]

        debt_equity = debt / equity

        return {
            "de_ratio_real": debt_equity,
            "cash_level": cash
        }
    except:
        return {}


def cashflow_factors(cashflow):

    try:
        fcf = cashflow.loc["Free Cash Flow"].iloc[0]
        op_cf = cashflow.loc["Operating Cash Flow"].iloc[0]

        quality = fcf / op_cf

        return {
            "fcf_quality": quality
        }
    except:
        return {}

def build_factor_row(symbol):

    data = load_yf_data(symbol)
    if data is None:
        return None

    factors = {}

    factors.update(price_factors(data["price"]))
    factors.update(income_factors(data["income"]))
    factors.update(balance_factors(data["balance"]))
    factors.update(cashflow_factors(data["cashflow"]))

    return factors

def build_factor_table(df):

    rows = {}

    for symbol in df.index:
        print("Processing", symbol)

        factors = build_factor_row(symbol)

        if factors:
            rows[symbol] = factors

    return pd.DataFrame.from_dict(rows, orient="index")

def normalize_factors(factors_df):

    scored = factors_df.copy()

    for col in scored.columns:
        scored[col] = scored[col].rank(pct=True)

    return scored
```

```python
FACTOR_WEIGHTS = {
    "momentum_12m": 0.25,
    "rev_growth_q": 0.15,
    "earn_growth_q": 0.15,
    "fcf_quality": 0.15,
    "de_ratio_real": -0.10,   # lower debt better
    "volatility": -0.10,      # lower risk better
    "cash_level": 0.10
}

def compute_quant_score(norm_df):

    score = pd.Series(0, index=norm_df.index)

    for factor, weight in FACTOR_WEIGHTS.items():
        if factor in norm_df:
            score += norm_df[factor] * weight

    return score

```

```python
def add_quant_score(df):

    factor_table = build_factor_table(df)

    norm_factors = normalize_factors(factor_table)

    scores = compute_quant_score(norm_factors)

    df["score"] = scores

    return df.sort_values("score", ascending=False)
```
